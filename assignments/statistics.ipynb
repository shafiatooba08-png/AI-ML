```python
from scipy import stats
import numpy as np
temps = [22, 25, 19, 30, 27, 21, 28, 24, 26, 23]
mean = np.mean(temps)
print(f'Mean score: {mean}')
median = np.median(temps)
print(f'Median: {median}')
result = stats.mode(temps)
print(f'Mode: {result.mode}')
```

```python
temps = [22, 25, 19, 30, 27, 21, 28, 24, 26, 23]
Range_data = max(temps) - min(temps)
a=np.var(temps) 
b=np.std(temps)
print(f'range_data:{Range_data}')
print(f'Variance: {a:.2f}')
mean = np.mean(temps)
std  = np.std(temps)
print(mean)
print(f'Standard Deviation: {std:.2f}')
print(f'\nRange  mean ± 1σ  = {mean-std:.1f}  to  {mean+std:.1f}')
print(f'Range  mean ± 2σ  = {mean-2*std:.1f}  to  {mean+2*std:.1f}')


```

```python
marks = [45, 52, 58, 61, 67, 72, 76, 80, 85, 91, 95, 99]
a=np.percentile(marks, 25)
b=np.percentile(marks, 50)
c=np.percentile(marks, 75) 
IQR = c-a
print("\n",a,"\n",b,"\n",c,"\n",IQR)
```

```python
salaries = [30000, 32000, 35000, 33000, 200000, 31000, 34000, 36000, 29000, 32500]
a=np.percentile(salaries,25)
b=np.percentile(salaries,75)
IQR= b-a
lower_fence=a-1.5*IQR
higher_fence=b+1.5*IQR
print(lower_fence)
print(higher_fence)
for s in salaries:
    if s<lower_fence or s>higher_fence:
       print(f's:{s}')
    
```

```python
import numpy as np
heights = np.array([150, 162, 170, 155, 180, 175, 168, 158, 172, 165])
a = np.mean(heights)
b = np.std(heights)
all_z_scores = (heights - a) / b
print(all_z_scores)
for h, z in zip(heights, all_z_scores):
    print(f"Height: {h} | Z-score: {z:.2f}")
```

```python
from scipy import stats
import numpy as np
right_skewed = [2, 3, 3, 4, 4, 4, 5, 5, 10, 15, 20]
left_skewed  = [1, 5, 10, 15, 16, 16, 17, 17, 18, 18, 19]
a=stats.skew(right_skewed)
b=stats.skew(left_skewed)
print(a)
print(b)
right_mean=np.mean(right_skewed)
right_median=np.median(right_skewed)
print(right_mean)
print(right_median)
left_mean=np.mean(left_skewed )
left_median=np.median(left_skewed )
print(left_mean)
print(left_median)
# 3. Formatted Output for Right Skewed
print("--- Right Skewed ---")
print(f"Skewness: {a:+.2f}   direction: positive →")
print(f"Mean: {right_mean:.2f}   Median: {right_median:.1f}")
if right_mean > right_median:
    print("Mean > Median ✓  (right skew rule holds!)")

print("\n--- Left Skewed ---") # \n adds a space between sections

# 4. Formatted Output for Left Skewed
print(f"Skewness: {b:.2f}   direction: negative ←")
print(f"Mean: {left_mean:.2f}   Median: {left_median:.1f}")
if left_mean < left_median:
    print("Mean < Median ✓  (left skew rule holds!)")
```
