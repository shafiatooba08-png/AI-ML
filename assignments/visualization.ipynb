```python
months     = ['Jan','Feb','Mar','Apr','May','Jun']
sales      = [200, 240, 180, 300, 350, 320]

cities     = ['Karachi','Lahore','Islamabad','Peshawar']
population = [16, 13, 2, 4]   # millions

hours      = [1, 2, 3, 4, 5, 6, 7, 8]
grades     = [50, 55, 60, 65, 72, 78, 85, 90]

categories = ['Food', 'Rent', 'Transport', 'Entertainment']
expenses   = [40, 30, 15, 15]

fig,axes=plt.subplots(2,2,figsize=(10,8))

axes[0,0].plot(months,sales,marker="o" )
axes[0,0].set_title('monthly sales')

axes[0,1].bar(cities,population)
axes[0,1].set_title('city population')

axes[1,0].scatter(hours,grades)
axes[1,0].set_title('grades per hours')

axes[1,1].pie(expenses,labels=categories)
axes[1,1].set_title('categories of expenses')
fig.suptitle('Monthly Dashboard',fontsize=14)
plt.tight_layout(rect=[0, 0, 1, 0.95], h_pad=4)
plt.show()
```

```python
from bokeh.plotting import figure, show
from bokeh.models import HoverTool, ColumnDataSource
from bokeh.io import output_notebook
output_notebook()
data = {
    'days': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'],
    'visitors': [120, 180, 150, 200, 250, 300, 270]
}
source = ColumnDataSource(data=data)
p = figure(x_range=data['days'], title="7-Day Website Traffic", x_axis_label='Day of the Week', y_axis_label='Number of Visitors',width=700, height=400)
p.line(x='days', y='visitors', source=source, line_width=2, color="navy", legend_label="Traffic")
p.circle(x='days', y='visitors', source=source, size=8, color="orange", fill_alpha=0.8)
hover = HoverTool(tooltips=[("Day", "@days"),("Visitors", "@visitors")])
p.add_tools(hover)
show(p)
```

```python
import plotly.express as px

# Load the built-in Gapminder dataset
df = px.data.gapminder()

# Create the animated scatter plot
fig = px.scatter(df, 
                 x="gdpPercap", 
                 y="lifeExp", 
                 animation_frame="year", 
                 animation_group="country",
                 size="pop", 
                 color="continent", 
                 hover_name="country", 
                 log_x=True,          # Apply log scale to the x-axis
                 size_max=55,         # Set maximum bubble size
                 range_x=[100, 100000], # Keep the x-axis steady during animation
                 range_y=[25, 90],    # Keep the y-axis steady during animation
                 labels={"gdpPercap": "GDP per Capita", "lifeExp": "Life Expectancy"},
                 title="GDP vs Life Expectancy Over Time")

# Show the plot
fig.show()
```

```python
import plotly.express as px

# Step 1: Load the Gapminder data
df = px.data.gapminder()

# Step 2: Filter for the year 2007 and the continent Asia
df_2007_asia = df[(df['year'] == 2007) & (df['continent'] == 'Asia')]

# Step 3: Sort the data by life expectancy in descending order
# This ensures our bar chart goes from highest to lowest automatically
df_2007_asia = df_2007_asia.sort_values(by='lifeExp', ascending=False)

# Step 4: Create the bar chart
fig = px.bar(df_2007_asia, 
             x='country', 
             y='lifeExp', 
             title="Life Expectancy in Asia (2007)",
             labels={'lifeExp': 'Life Expectancy (Years)', 'country': 'Country'},
             color='lifeExp',  # Adds a color gradient based on the value
             color_continuous_scale='Viridis')

fig.show()
```
