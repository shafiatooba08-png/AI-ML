```python
v1=np.array([1,0])
degree=90
rad=np.radians(degree)
#R = np.array([[cosθ, -sinθ],[sinθ, cosθ]])
c=np.cos(rad)
s=np.sin(rad)
# 2. Build the Rotation Matrix R
# R = [[cosθ, -sinθ], [sinθ, cosθ]]
R = np.array([[c, -s],
              [s,  c]])

# 3. Rotate the vector using the @ operator (Matrix Multiplication)
v_rotated = R @ v1

print(f"Original Vector: {v1}")
print(f"Rotation Matrix:\n{np.round(R,1)}")
print(f"Rotated Vector: {np.round(v_rotated,1)}")
```
