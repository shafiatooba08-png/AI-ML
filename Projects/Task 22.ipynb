# Valid and Invalid Dictionaries

Part A: Key must be IMMUTABLE

Type : String "name" , Integer 1 , Tuple (1,2)

List [1,2] cannot be key value 

✅ Valid - integer as key
my_dict = {1: "one", 2: "two"}

✅ Valid - tuple as key
my_dict = {(1, 2): "one two"}

 ❌ Invalid - list as key (ERROR aayega!)
my_dict = {[1, 2]: "hello"}

 ✅ Valid - string key, list value (value kuch bhi ho sakti hai!)
my_dict = {"cities": ["Lahore", "Karachi", "Islamabad"]}

# Access Dictionary Items

```python
student = {"name": "Shafia", "city": "Lahore", "hobby": "reading"}
print(student["name"]) # Square brackets - KEY NA MILE toh ERROR aayega!
print(student.get("1")) # get() - KEY NA MILE toh None return karta hai
```

```python
phone = {"brand": "Samsung","model": "A54","price": 80000,"city": "Lahore"}
print(phone["brand"])
print(phone.get("price"))
print(phone.get("Colour","not found"))
```

 # Add Items to a Dictionary

```python
laptop = {"brand": "Dell", "ram": "8GB"}
laptop["storage"]="512GB" # ek item add
laptop.update({"price":50000,"color":"black"}) # multiple items ek saath
print(laptop)
```

# Remove Items

```python
bag = {"book": "Python", "pen": "blue", "notebook": "A4", "eraser": "white"}
del bag["eraser"] #specific item delete
print(bag)
removed=bag.pop("pen") # specific item delete but return value 
print(removed)
bag.clear() #sab delete
print(bag)
```

# Change Dictionary Items


```python
employee = {"name": "Shafia", "company": "Trivia","role": "Teacher","salary": 30000}
employee["role"]= "AI Engineer"
employee.update({"company":"Netsol","salary":2000000})
print(employee)
```

 # Iterate Through a Dictionary

```python
marks = {"Python": 95, "ML": 88, "Statistics": 92, "Linux": 85}
for key in marks:
    print(key ,":",marks[key])
for value in marks.values():
    print(value)
for key, value in marks.items():
    print(key ,':', value)
```

# Find Dictionary Length

```python
bag = {"laptop": "Dell","charger": "60W","notebook": "A4","pen": "blue","bottle": "1L"}
print(len(bag))
bag['pencil']=2
print(len(bag))
del bag['pen']
print(len(bag))
```

# Python Dictionary Methods

```python
city_info = {"name": "Lahore", "province": "Punjab","population": 1400000,"famous_for": "food"}
print(city_info.keys())
print(city_info.values())
copy=city_info.copy()
copy["population"]="1000000"
last= city_info.popitem() # remove key,value and return them in the form of tuple 
print(city_info )
print(copy)
print(last)
```
