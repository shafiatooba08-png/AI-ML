# Multiline String

```python
import smtplib
from email.mime.text import MIMEText

email_body = '''Dear Customer,

Congratulations! Your loan has been approved.
Amount: PKR 500,000
Tenure: 36 months

Regards,
NETSol Finance Team'''

# email details
sender = 'shafiatooba08@gmail.com'
receiver = 'studentbscs2022@gmail.com'
password = 'kopyojmaqrbmzxrs'

# create the email
msg = MIMEText(email_body)
msg['Subject'] = 'Loan Approval Notification'
msg['From'] = sender
msg['To'] = receiver

# send the email
with smtplib.SMTP_SSL('smtp.gmail.com', 465) as server:
    server.login(sender, password)
    server.sendmail(sender, receiver, msg.as_string())
    print("Email sent successfully!")
```

# String Length 

```python
message = '''Lorem ipsum dolor sit amet, consectetur adipiscing elit. 
Sed do eiusmod tempor incididunt ut labore et dolore magna aliqua. 
Ut enim ad minim veniam quis nostrud exercitation.'''
if len(message)> 160:
    calculation= len(message)-160
    print(f'{calculation} characters over limit')
else:
    print('Message Sent ✅')
print(len(message))
```

 # String Membership Test

```python
message = "Hello, how are you?"
Banned_words=["spam" ,"hack", "virus"]
for i in Banned_words:
    if i in message:
       print('Message Blocked ❌ — contains banned word') 
       break
else:
     print('Message Sent ✅')
```

# lower()

```python
user1 = "ALI_123"
user2 = "ali_123"
a=user1.lower()
b=user2.lower()
if a==b:
    print('Same User ✅')
else:
    print('Different User ❌')
```

# replace()

```python
message = "this is spam and virus message"
Banned_words=["spam" ,"hack", "virus"]
for i in Banned_words:
    if i in message:
        message = message.replace(i, "***")
print(message) 
       
```
