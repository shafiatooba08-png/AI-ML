# What is a DataFrame?

```python
import pandas as pd 
data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}
df=pd.DataFrame(data)
print(df)
```

# Locate Row


Pandas use the loc attribute to return one or more specified row(s)

```python
print(df.loc[0])
```

```python
print(df.loc[[0,1]])
```

df.loc[0] — Result: One row (looks like a list of values).

df.loc[[0, 1]] — Result: A new table containing only row 0 and row 1.

# Named Indexes

```python
data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}
df=pd.DataFrame(data,index=["x","y","z"])
print(df)
```

# Locate Named Indexes

```python
print(df.loc["x"])
```
