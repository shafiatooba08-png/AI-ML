# Tokenization

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize, sent_tokenize

text = "I love eating biryani in Lahore. It is my favorite food!"

word_tokens = word_tokenize(text)
print("Word Tokens:", word_tokens)

sent_tokens = sent_tokenize(text)
print("Sentence Tokens:", sent_tokens)
```

# Removal of stopwords

```python
import nltk
nltk.download('stopwords')

from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
print(stop_words)
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

text = "I love eating biryani in Lahore. It is my favorite food!"

# Pehle tokenize karo
tokens = word_tokenize(text)
print("Before:", tokens)

# Stopwords hatao
stop_words = set(stopwords.words('english'))

filtered_tokens = [word for word in tokens if word.lower() not in stop_words]
print("After:", filtered_tokens)
```

# remove punction 

```python
import string

filtered_tokens = [word for word in tokens 
                   if word.lower() not in stop_words 
                   and word not in string.punctuation]

print("Clean tokens:", filtered_tokens)
```

# Stemming

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

words = ['eating', 'eaten', 'eats', 'loved', 'running', 'happily', 'studies', 'finally']

for word in words:
    print(f"{word} → {ps.stem(word)}")
```

# Lemmatization

```python
import nltk
nltk.download('wordnet')

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ['eating', 'eaten', 'eats', 'loved', 'running', 'happily', 'studies', 'finally']

for word in words:
    print(f"{word} → {lemmatizer.lemmatize(word)}")
```

```python
for word in words:
    print(f"{word} → {lemmatizer.lemmatize(word, pos='v')}")
#                                                   ↑
#                                              'v' = verb
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import string

def nlp_pipeline(text):
    # Step 1: Tokenization
    tokens = word_tokenize(text)
    print("Step 1 - Tokens:", tokens)
    
    # Step 2: Stopword + Punctuation Removal
    stop_words = set(stopwords.words('english'))
    tokens = [w for w in tokens if w.lower() not in stop_words 
              and w not in string.punctuation]
    print("Step 2 - Clean Tokens:", tokens)
    
    # Step 3: Lemmatization
    lemmatizer = WordNetLemmatizer()
    tokens = [lemmatizer.lemmatize(w, pos='v') for w in tokens]
    print("Step 3 - Lemmatized:", tokens)
    
    return tokens

# Test karo!
text = "I love eating biryani in Lahore. It is my favorite food!"
result = nlp_pipeline(text)
print("\nFinal Result:", result)
```
