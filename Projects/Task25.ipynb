# NumPy Creating Arrays

```python
import numpy as np
arr=np.array([1,2,3,4,5,6])
print(arr)
print(type(arr))
```

```python
tuples=np.array((1,2,3,4,5))
print(tuples)
print(tuples.ndim)
```

```python
array=np.array(5)
print(array)
print(array.ndim)
```

```python
arr=np.array([1,2])
print(arr)
print(arr.ndim)
```

```python
arr=np.array([[1,2],[3,4]])
print(arr)
print(arr.ndim)
```

```python
arr1=np.array([[[1,2],[3,4]],[[5,6],[7,8]],[[8,10],[11,12]]])
print(arr1)
print(arr1.ndim)
```

```python
arr0=np.array([1,2,3,4,5],ndmin=8)
print(arr0)
print(arr0.ndim)
```
