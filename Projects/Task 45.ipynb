# Pandas - Data Correlations

The corr() Method
The corr() method calculates the correlation coefficient between every pair of numeric columns in a DataFrame.

It ignores non-numeric columns (like strings) automatically.

```python
import pandas as pd 
df=pd.read_csv("Downloads/health_data.csv")
print(df.corr(numeric_only=True))
```

Relationship Types:

1.0: Perfect relationship (positive).

0.9: Strong positive relationship.

-0.9: Strong negative relationship (one goes up, the other goes down).

0.2 / 0.0: Weak or no relationship.
