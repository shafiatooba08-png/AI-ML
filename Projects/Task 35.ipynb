# NumPy Searching Arrays

```python
import numpy as np 
arr = np.array([1, 2, 3, 4, 5, 4, 4])
x=np.where(arr==4)
print(x)
```

# Find the indexes where the values are odd:

```python
arr = np.array([10, 14, 93, 41, 8, 7])
x=np.where(arr%2==1)
print(x)
```

# Search Sorted

```python
arr = np.array([6, 7, 8, 9])
x=np.searchsorted(arr,7)
print(x)
```

# Search From the Right Side

```python
arr = np.array([6, 7, 8, 9])
x=np.searchsorted(arr,7,side="right")
print(x)
```

# Multiple Values

```python
arr = np.array([1, 3, 5, 7])
x=np.searchsorted(arr,[2,4,6])
print(x)
```

We want to insert the number 2:

Compare: Is 2 smaller than 1 (Index 0)? No.

Compare: Is 2 smaller than 3 (Index 1)? Yes.

The Action: We take Index 1.

The Slide: The numbers 3, 5, and 7 all "slide right" to indices 2, 3, and 4.
