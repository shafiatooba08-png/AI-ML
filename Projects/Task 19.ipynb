# Access Tuple Items

```python
customer = ('Ali Hassan', 'AC-00456', 'Current', 'KHI-02')
print(f'Welcome , {customer[0]}!\nAccount Type: {customer[-2]}\nYour Branch Code: {customer[-1]}')
```

# Tuple Length

```python
doctors = ('Dr. Ahmed', 'Dr. Sara', 'Dr. Bilal', 'Dr. Ayesha', 'Dr. Usman')
print(f'Total doctors available today: {len(doctors)}')
```

# Iterate Through a Tuple

```python
witnesses = ('Kamran Ali', 'Sana Mir', 'Tariq Hussain', 'Reema Khan')
for index, value in enumerate(witnesses,1):
    print(f'Witness {index}: {value}')
```

# Check if Item Exists in a Tuple

```python
blood_bank = ('A+', 'B+', 'O+', 'AB+', 'O-')
group=input('Enter required blood group: ')
if group in blood_bank:
    print(f'{group} is available in blood bank')
else:
    print(f'{group} is not available in blood bank. Emergency request sent!')
```

# Change Tuple Items (Immutability)

```python
employee = ('E-1021', 'Ali Khan', 'Lahore', 'Junior Analyst')

temp = list(employee)

# automatically find the position of 'Junior Analyst'
position = temp.index('Junior Analyst')

temp[position] = 'Senior Analyst'
employee = tuple(temp)
print(position)
print(f'Updated Record: {employee}')
```

# Delete a Tuple

```python
session = ('user_101', 'token_xk92z', '192.168.1.5')
del session

try:
    print(session)
except NameError:
    print("Session terminated. No active session found.")
```
