# Using enumerate()

```python
transactions = ['Deposit 5000', 'Withdrawal 2000', 'Transfer 15000', 'Deposit 3000', 'Withdrawal 500']
for index,value in enumerate(transactions,1):
    print(f'TXN-{index}: {value}')
```

```python
login_attempts = ['admin', 'user_42', 'root', 'guest', 'DROP TABLE users', 'alice']
for index,value in enumerate(login_attempts,1):
    if len(value.split()) > 1:                             #" " in value:      "DROP TABLE users".split()  →  ['DROP', 'TABLE', 'users'] 
       print(f'⚠ Attempt {index} is SUSPICIOUS: {value}')
    else:
         print(f'Attempt {index} OK: {value}')
    
    
```

# Using zip()

```python
account_holders = ['Ali', 'Sara', 'Ahmed', 'Hina', 'Usman']
balances = [15000, 800, 23000, 450, 5000]
for name ,balance in zip(account_holders,balances):
    if balance<1000:
        print(f'{name} - Balance: {balance} ⚠ LOW BALANCE')
    else:
         print(f'{name} - Balance: {balance} ✅')
```

```python
ip_addresses = ['192.168.1.1', '10.0.0.5', '172.16.0.3', '192.168.1.9', '10.0.0.8']
packet_counts = [200, 1500, 800, 2300, 450]
for ip ,packet in zip(ip_addresses,packet_counts):
    if packet>1000:
       print(f'{ip} - packet: {packet} 🚨 DDoS ALERT')
    else:
        print(f'{ip} - packet: {packet} ✅ Normal')
```

# Using items()

```python
suspects = {
    'Kamran Shah': 'Cleared',
    'Ali Raza': 'Wanted',
    'Nadia Malik': 'Under Investigation',
    'Zafar Iqbal': 'Wanted',
    'Sara Ahmed': 'Cleared'
}
for key,value in suspects.items():
    if value == 'Wanted':
       print(f'{key} - Status: {value} 🚨 ARREST IMMEDIATELY')
    else:
         print(f'{key} - Status: {value} ✅')
    
```

```python
servers = {
    'Web-Server-01': 3,
    'Database-Server': 9,
    'Mail-Server': 6,
    'Auth-Server': 8,
    'Backup-Server': 2
}
for key,value in servers.items():
    if value <7:
        print(f'{key} - Score: {value}  ✅ Safe')
    else:
         print(f'{key} - Score: {value}  🚨 CRITICAL - PATCH IMMEDIATELY')
```

# Using sorted()

```python
transactions = [45000, 1200, 890000, 3400, 67000, 250, 120000]
for i in sorted(transactions):
     print(f'Transaction Amount: {i}')
```

```python
open_ports = [8080, 443, 22, 3306, 80, 22, 443, 21, 8080]
for i in sorted(set(open_ports)):
    print(f'Priority Port: {i}')
```

# While Loop with If Condition

```python
import psutil
while True: # Run forever — no condition to stop it. It just keeps looping endlessly until something inside forces it to stop.
      cpu_usage =  psutil.cpu_percent(interval=1)
      ''' Every loop cycle — go ask the operating system "what is CPU usage right now?" and wait 1 second before getting the answer. Store that number in cpu_usage.
      The interval=1 is what makes it check every second — like a heartbeat.'''

      print(f'🔍 Monitoring... CPU Usage:{cpu_usage}')                                       
      if cpu_usage>=90:
         print('🚨 CRITICAL! CPU at 91% - Killing Malware Process!')
         break
    
```
