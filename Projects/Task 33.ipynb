# Joining NumPy Arrays

```python
import numpy as np 
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
arr=np.concatenate((arr1,arr2))
print(arr)
    
```

# Join two 2-D arrays along rows (axis=1):

Axis 0: Makes the array Longer (more rows).

Axis 1: Makes the array Wider (more columns).

```python
arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])
arr=np.concatenate((arr1,arr2),axis=1)
print(arr)
```

# Joining Arrays Using Stack Functions

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
arr=np.stack((arr1,arr2),axis=0)
print(arr)
```

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
arr=np.concatenate((arr1,arr2),axis=0)
print(arr)
```

# Stacking Along Rows

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
arr=np.hstack((arr1,arr2))
print(arr)
```

# Stacking Along Columns

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
arr=np.vstack((arr1,arr2))
print(arr)
```

# Stacking Along Height (depth)

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
arr=np.dstack((arr1,arr2))
print(arr)
```
