```python
try:
    CNIC=input("please enter your CNIC number")
    if len(CNIC)==13 and CNIC.isdigit():
        print("CNIC Verified! Record found.")
    else:
         print("Invalid CNIC! Must be 13 digits.")
except:
      print("Invalid CNIC! Only numbers allowed.")
     
```

# Specific Exceptions

```python
try:
    IP = int(input("enter your last 4 digit of ip adress "))
    divide=10/IP
    print(divide)
except ValueError:
      print("Invalid! IP address sirf numbers mein hota hai.")
except ZeroDivisionError:
      print("ip not divisible by 0")
```

# else Clause

```python
try:
    port=int(input("please enter port"))
except ValueError:
       print("Invalid! Port number must be numbers")
else: 
     print("Port number valid hai! Scanning port:",port)
```
