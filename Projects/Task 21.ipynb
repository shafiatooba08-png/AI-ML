# Create a Set

```python
attendance = [112, 114, 112, 116, 114, 118, 115]
a=set(attendance)
print(f'unique students: {a}')
print(f'Total count : {len(a)}')
```

# Create an Empty Set

```python
customers=set()
customer_data={}
customers.add(1001)
print(type(customers))
print(type(customer_data))
print(customers)
```

# Duplicate Items in a Set

```python
products_sold=set()
product_IDs = [101 ,205 ,101 ,307 ,205 ,101]
for i in product_IDs:
    products_sold.add(i)
    print(products_sold)
```

```python
monday = {101, 205, 307}
monday.update([101, 205, 408, 509, 408])
print(monday)
```

# Remove an Element from a Set

```python
languages = {'Python', 'Java', 'C++', 'JavaScript', 'Swift'}
languages.discard('Java')
print(languages)
languages.remove('C++')
print(languages)
```

# Built-in Functions with Set

```python
numbers = {15, 3, 8, 42, 7, 23}
print(len(numbers))
print(max(numbers))
print(min(numbers))
print(sum(numbers))
print(type(sorted(numbers)))
```

# Iterate Over a Set in Python

```python
fruits = {"Mango", "Apple", "Peach", "Banana", "Grapes"}
for i in fruits:
    if len(i)>5:
       print(i)
```

 # Find Number of Set Elements

# Python Set Operations

```python
A = {'Ali', 'Sara', 'John', 'Ayesha'}
B = {'Sara', 'Ayesha', 'Usman', 'Zara'}
#print(A.union(B))
print(A|B)
print(A&B)
# print(A.intersection(B))
print(A-B)
#print(A.difference(B))
print(A^B)
#print(A.symmetric_difference(B))
```

# Other Python Set Methods

```python
numbers = {1, 2, 3, 4, 5}
numbers.clear()
print(numbers)
```

```python
original = {1, 2, 3, 4, 5}
copy=original.copy()
original.add(99)
print('original:',original)
print('copy:',copy)
```

```python
original = {1, 2, 3, 4, 5}
copied = original  # copy() nahi!
original.add(99)

print('Original:', original)
print('Copied:', copied)
```
