# inplace = True argument:

inplace=False: You click "Save As"

inplace=True: You just click "Save".

```python
df=pd.read_csv("Downloads/health_data.csv")
df.dropna(inplace=True)
print(df.to_string())
```

# Replace Empty Values

Another way of dealing with empty cells is to insert a new value instead.

This way you do not have to delete entire rows just because of some empty cells.

The fillna() method allows us to replace empty cells with a value:

```python
df=pd.read_csv("Downloads/health_data.csv")
df.fillna(130,inplace=True)
#print(df.to_string())
print(df)
```

# Replace Only For Specified Columns

```python
df=pd.read_csv("Downloads/health_data.csv")
df.fillna({"Calories":130},inplace=True)
print(df)
```

# Removing Rows

```python
df.dropna(subset=['Date'], inplace = True)
print(df)
```

Result: Only Row 22 is deleted.

Why?: Because you told Pandas to only care about the Date column.

Row 18 survives because even though its Calories are missing, its Date is perfectly fine.
