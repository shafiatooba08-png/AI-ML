# Sorting Arrays

# Sort the array:

```python
import numpy as np 
arr = np.array([3, 2, 0, 1])
print(np.sort(arr))
```

# Sort the array alphabetically:

```python
arr = np.array(['banana', 'cherry', 'apple'])
print(np.sort(arr))
```

# Sort a boolean array:

```python
arr = np.array([True, False, True])
print(np.sort(arr))
```

# Sorting a 2-D Array

```python
arr = np.array([[3, 2, 4], [5, 0, 1]])
print(np.sort(arr)) 
```
