```python
import pandas as pd
a=[1,2,3,4,5]
x=pd.Series(a)
print(x)
```

Key Characteristics of a Series:
One-Dimensional: It is a single list of values.

Labeled: Every value in a Series has a label called an index.

Homogeneous Data: While it can hold any type,
a single Series usually holds the same type of data (all integers, all strings, etc.) to make calculations faster.

# Labels=index

```python
print(x[0])
```

# Create Labels

```python
a = [1, 7, 2]
x=pd.Series(a,index=["x","y","z"])
print(x)
```

```python
print(x["z"])
```

# Key/Value Objects as Series

```python
calories = {"day1": 420, "day2": 380, "day3": 390}
y=pd.Series(calories)
print(y)
```

The keys of the dictionary become the labels.

# Create a Series using only data from "day1" and "day2":

```python
calories = {"day1": 420, "day2": 380, "day3": 390}
y=pd.Series(calories,index=["day1","day2"])
print(y)
```

# DataFrames

```python
data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}
df=pd.DataFrame(data)
print(df)
```
