```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Text
text = "I am going to the market to buy some fruits and vegetables"

# Stopwords load karo
stop_words = stopwords.words('english')

# Tokenize karo
tokens = word_tokenize(text)

# Stopwords remove karo
filtered_tokens = [word.lower() for word in tokens if word.lower() not in stop_words]

# Compare karo
print("Original text:", text)
print("\nTotal words:", len(tokens))
print("After removing stopwords:", len(filtered_tokens))
print("\nBefore:", tokens)
print("After:", filtered_tokens)
```

# SpaCy Method

```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = "I am going to the market to buy some fruits and vegetables"

doc = nlp(text)

filtered_words = [token.text for token in doc if not token.is_stop]

print("Before:", text)
print("\nAfter:", filtered_words)
```

# Gensim Method

```python
from gensim.parsing.preprocessing import remove_stopwords

text = "I am going to the market to buy some fruits and vegetables"

filtered_text = remove_stopwords(text)

print("Before:", text)
print("\nAfter:", filtered_text)
```

# Custom Stopwords

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

stop_words = set(stopwords.words('english'))

custom_words = ['market', 'buy', 'fruits']

stop_words.update(custom_words)

text = "I am going to the market to buy some fruits and vegetables"
tokens = word_tokenize(text)

filtered = [word.lower() for word in tokens if word.lower() not in stop_words]

print("Custom stopwords added:", custom_words)
print("\nBefore:", tokens)
print("\nAfter:", filtered)
```
